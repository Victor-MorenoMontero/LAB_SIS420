## Función sigmoid

```python
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))
```

```python
print(sigmoid(0))
print(sigmoid(1))
print(sigmoid(-1))
```

## Función de costo para regresión logística

```python
def lrCostFunction(theta, X, y, lambda_):
    m = y.size

    h = sigmoid(X.dot(theta))

    J = (1 / m) * np.sum(
        -y.dot(np.log(h))
        - (1 - y).dot(np.log(1 - h))
    )

    reg = (lambda_ / (2 * m)) * np.sum(theta[1:] ** 2)

    J = J + reg

    return J
```

## Función para calcular el gradiente

```python
def lrGradient(theta, X, y, lambda_):
    m = y.size

    h = sigmoid(X.dot(theta))

    grad = (1 / m) * X.T.dot(h - y)

    grad[1:] = grad[1:] + (lambda_ / m) * theta[1:]

    return grad
```
